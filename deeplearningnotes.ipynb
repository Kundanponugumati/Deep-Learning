A PyTorch tensor can:
- perform mathematical operations
- calculate gradients
- run on GPU
- participate in neural-network computation

```python
import torch
x = torch.tensor(5.0, requires_grad=True)

y = x ** 2

y.backward()

print(x.grad)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

c = np.dot(a, b)
print(c)
print(c.shape)
```

```python
A = np.array([
    [1,2,3],
    [4,5,6]
])

B = np.array([
    [1,2],
    [3,4],
    [5,6]
])
print("A dot of B \n")
print(A.dot(B))
print(A.dot(B).shape)
print("A @ B \n")
print(A @ B)
print((A @ B).shape)
```
