### reshape()

```python
x = torch.tensor([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])

print(x.shape)
```

```python
y = x.reshape(4, 2)

print(y)
print(y.shape)
```

- fundamental rule when reshaping no of elements should remain same. 

```python
y = x.reshape(2, -1)
print(y)
print(y.shape)
# what -1 means it is a question mark like 2 * ? = 8. so ? = 4. that what happened here. 
```

```python
y = x.reshape(-1,8)
print(y)
print(y.shape)
```

### unsqueeze()

```python
x = torch.tensor([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])
print(x)
print(x.shape)
print(x.ndim)
```

```python
y = x.unsqueeze(0)
# we are adding another dimension at position 0
print(y)
print(y.shape)
print(y.ndim)
```

```python
y = x.unsqueeze(1)
# we are adding another dimension at position 1
print(y)
print(y.shape)
print(y.ndim)
```

### sequeeze()

```python
# here we remove the any position that contains dimension as 1
"""
suppose if you are having 
[1,3,1] -> [3]
[2,3,1] -> [2,3]
[3,4] -> [3,4] no effect becasue we are not having dimension 1
"""
```

```python
torch.manual_seed(42)
x_tensor = torch.randint(low = 0, high = 11,size=[1,3,1,2],dtype=torch.int64)
print(x_tensor)
print(x_tensor.shape)
print(x_tensor.ndim)
```

```python
y = x_tensor.squeeze()
print(y)
print(y.shape)
print(y.ndim)
```

### transpose()

```python
# core thing transpose(0,1) means we are swapping 0 dimension with 1 and 1 dimension with 0
y = x_tensor.transpose(2,3) # means we are swapping 2 and 3 dimensions so output of shape will be [1,3,2,1]
print(y)
print(y.shape)
print(y.ndim)
```

### permute() 

```python
# similar to transpose but in transpose we can only shape 2 dimensions but if you change multiple dimensions we use permute()

print(y)
print(y.shape)
print(y.ndim)
```

### cat()

```python
# we will concatinate the dimension
y1 = x_tensor.permute(2,0,3,1)
y2 = x_tensor.permute(2,1,3,0)
print(y1.shape,y1.ndim)
print(y2.shape,y2.ndim)
torch.cat(y1,y2)
torch.stack(y1,y2)
```

### multiplication

```python
torch.manual_seed(42)
A = torch.randint(low = 0, high = 11, size= [2,2],dtype=torch.int64)
B = torch.randint(low = 0, high = 11, size= [2,2],dtype=torch.int64)
print(A,"\n",B)
print(A @ B)
print(A * B)
```

```python
a = torch.tensor([1,3,4])
b = torch.tensor([[1,3,4],[4,5,6],[7,8,9]])
print(a.shape,b.shape)
print(a,"\n",b)
print(a @ b)
print(a * b)
```
