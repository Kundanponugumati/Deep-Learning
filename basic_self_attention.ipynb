## Simple self attention

```python
import torch

X = torch.tensor([
    [1.0, 0.0, 1.0, 0.0],  # I
    [0.0, 2.0, 0.0, 1.0],  # love
    [1.0, 1.0, 0.0, 2.0]   # cats
])

print(X)
print(X.shape)
```

```python
print(X)
```

```python
Q = X
K = X
V = X
```

```python
scores = Q @ K.T

print(scores)
```

```python
d_k = K.shape[-1]

scores = scores / torch.sqrt(torch.tensor(d_k))

print(scores)
```

```python
attention_weights = torch.softmax(scores, dim=-1)

print(attention_weights)
```

```python
output = attention_weights @ V

print(output)
```
