### WTF Transformer


```python
Text
 ↓
Tokenizer
 ↓
Token IDs
 ↓
Token Embeddings + Positional Embeddings
 ↓
┌─────────────────────────────┐
│ Transformer Block           │
│                             │
│ Multi-Head Self-Attention   │
│        ↓                    │
│ Residual + LayerNorm        │
│        ↓                    │
│ Feed Forward Network        │
│        ↓                    │
│ Residual + LayerNorm        │
└─────────────────────────────┘
        ↓
(repeat N times)
        ↓
Linear projection to vocabulary
        ↓
Softmax / next-token prediction
```

```python
# Data
text = """
the cat sat on the mat
the dog sat on the floor
the cat ate the food
the dog ate the food
"""
```

```python
# splitting data for tokens.

words = text.split()
vocab = sorted(set(words))

print(vocab)
print(len(vocab))
```

```python
# LLM can't work with words. we need ID's
# String to Integer
stoi = {word:i for i,word in enumerate(vocab)}
print(stoi)
```

```python
# Integer to String
itos = {i: word for word, i in stoi.items()}
print(itos)
```

```python
# now we need to encode the full thing. 

encoded = [stoi[word] for word in words]
print(encoded)
```

```python
# we need to encode each line. 
encode = [stoi[word] for word in text.split()]
print(encode)
```
